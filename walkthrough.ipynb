## 2. Variables and Types

A variable is a name that stores a value. Python has a few basic types: `int` (whole numbers), `float` (decimals), `str` (text), and `bool` (`True`/`False`).

```python
age = 16
height_in_meters = 1.7
name = "Ada"
likes_python = True

print(name, age, height_in_meters, likes_python)
print(type(age), type(height_in_meters), type(name), type(likes_python))
```

You can do math with numbers (`+ - * /`) and combine strings with `+` or f-strings like `f"{name} is {age}"`.

```python
print(age + 4)
print(f"{name} will be {age + 4} in 4 years.")
```

## 3. Conditionals

`if` / `elif` / `else` let your program make decisions. Comparisons like `==`, `>`, `<`, `>=` give back `True` or `False`.

```python
temperature = 85

if temperature > 90:
    print("It's really hot.")
elif temperature > 70:
    print("It's warm.")
else:
    print("It's cool.")
```

## 4. Lists and Dictionaries

A `list` is an ordered collection of values. A `dict` (dictionary) stores key-value pairs.

```python
fruits = ["apple", "banana", "cherry"]
print(fruits[0])      # first item
fruits.append("date")  # add to the end
print(fruits)
print(len(fruits))     # how many items

student = {"name": "Ada", "age": 16, "likes_python": True}
print(student["name"])
student["age"] = 17
print(student)
```

## 5. Loops

A `for` loop repeats code once per item in a collection. A `while` loop repeats as long as a condition is `True`.

```python
for fruit in fruits:
    print(f"I like {fruit}")

count = 0
while count < 3:
    print(f"count is {count}")
    count += 1
```

## 6. Functions

A function is a reusable block of code. You define it once with `def`, then call it as many times as you want.

```python
def greet(person_name):
    return f"Hello, {person_name}!"

print(greet("Ada"))
print(greet("Sam"))
```
